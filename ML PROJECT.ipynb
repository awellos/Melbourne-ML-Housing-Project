Testing the assumptions of linear regression

1.Linearity

```python
import seaborn as sns
sns.set(style='ticks',color_codes=True,font_scale=2)
g=sns.pairplot(mh1_filled,height=3,diag_kind='hist',kind='reg')
g.fig.suptitle('Scatter Plot',y=1.08)
```

2.Multicollinearity

```python
corr=mh1_filled.corr()
print(corr)
```

3.Homoscedasticity

```python
from statsmodels.formula.api import ols
reg=ols('Price~Bedroom2+Bathroom+Rooms+Car+Distance+Suburb_cat+Type_cat+CouncilArea_cat+Propertycount+Address_cat+Method_cat+SellerG_cat+Postcode+Landsize+Lattitude+Longtitude+Regionname_cat',data=mh1_filled).fit()
print(reg.summary())
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(8,3.5))
pred_val=reg.fittedvalues.copy()
true_val=mh1_filled['Price'].values.copy()
resid=true_val-pred_val
sns.residplot(resid,pred_val)
plt.title('Homoscedasticity')
```

4.Normality of Residuals

```python
import scipy.stats as stats
fig,ax=plt.subplots(figsize=(10,6))
stats.probplot(resid,dist='norm',plot=plt)
```
